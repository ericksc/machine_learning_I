## **Contexto del Problema**

Se dispone de un conjunto de datos que describe **características físicas de diferentes frutas**, como su masa, altura, ancho y color promedio.  
El objetivo es construir un modelo de **clasificación automática** que identifique el tipo de fruta según estas dimensiones.


# **Parte 1 – Comparación de Clasificadores Tradicionales**

### Objetivo
Evaluar el rendimiento de diferentes algoritmos de clasificación (por ejemplo, Árbol de Decisión, K-Nearest Neighbors, Regresión Logística) para determinar cuál se adapta mejor al conjunto de datos de frutas.

###  Instrucciones
1. Cargue el dataset proporcionado y realice una exploración inicial de las variables.  
2. Prepare los datos (limpieza, selección de características y **escalado si es necesario**).  
3. Entrene al menos **tres clasificadores distintos** de scikit-learn.  
4. Aplique **validación cruzada (cross validation)** para medir el rendimiento promedio de cada modelo.  
5. Registre las métricas obtenidas (**accuracy, precision, recall, F1**) y preséntelas en una **tabla o gráfico comparativo**.  
6. Determine cuál clasificador presenta el mejor desempeño y redacte una **breve conclusión** justificando su resultado.

###  Resultados Esperados
- Tabla o gráfico con las métricas promedio de cada modelo.  
- Análisis comparativo del rendimiento.  
- Conclusión sobre el modelo más adecuado.


#  **Parte 2 – Clasificación con Red Neuronal (Keras)**

###  Objetivo
Implementar una **Red Neuronal Artificial** con **Keras (TensorFlow)** para clasificar las frutas y **comparar su rendimiento** con los modelos tradicionales de la Parte 1.

###  Instrucciones
1. Utilice el **mismo conjunto de datos y preprocesamiento** de la Parte 1.  
2. Defina una red neuronal **secuencial** con:
   - Capa de entrada adaptada al número de variables.  
   - **Dos capas ocultas** con funciones de activación no lineales (por ejemplo ReLU).  
   - **Capa de salida** con activación *softmax* para clasificación multiclase.  
3. Compile el modelo especificando:
   - **Optimizador** (Adam u otro).  
   - **Función de pérdida** adecuada para clasificación categórica.  
   - **Métricas** como *accuracy*.  
4. Entrene la red neuronal utilizando **validación durante las épocas** para observar la evolución del error y la exactitud.  
5. Evalúe el modelo con el conjunto de prueba y obtenga:
   - **Accuracy, Precision, Recall, F1-Score** y **Matriz de Confusión**.  
6. Presente los resultados en **tablas y gráficos**.  
7. Compare el rendimiento del modelo neuronal con los clasificadores de la Parte 1.  
8. Elabore una **conclusión** que resuma los hallazgos, discutiendo ventajas y posibles mejoras.

###  Resultados Esperados
- **Curvas de entrenamiento** (loss / accuracy).  
- **Tabla comparativa** con métricas de todos los modelos.  
- Conclusión sobre cuál clasificador obtuvo mejor desempeño.
