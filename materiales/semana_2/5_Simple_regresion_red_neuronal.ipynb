## Objetivo:

Implementar una red neuronal para que aprenda la relación entre 2 variables de datos númericos para predecir cual sería la salida para nuevos valores.

Se van a realizar los siguiente pasos:

- Definición de los datos
- Minería de datos: Aplicación de modelo de regresión lineal
- Predición

# Pasos

## Importación de las bibliotecas

```
# se importan las blibliotecas de python que se van usar
import matplotlib.pyplot as plt

import tensorflow as tf
tf.keras.utils.set_random_seed(42)
import numpy as np
import pandas as pd
from tensorflow import keras

import plotly.express as px
import plotly.graph_objects as go
```

## Definición de los datos

Se contruye un set de datos con información arbitraria.
- X para las entradas.
- Y para las salidas.

```
datos_df = pd.DataFrame({
    'x': [-1, 0, 1, 2, 3, 4],
    'y':[-3, -1, 1, 3, 5, 7]
     }
)

datos_df.head()
```

## Caracterización

```
fig = px.scatter(datos_df, y="y", x="x")
fig.update_traces(marker_size=10)
fig.show()
```

# Creación de la red neuronal

```
# Se crea un modelo usando un red neuronal de una sola neurona
red_modelo = keras.Sequential([keras.layers.Dense(units=1, input_shape=[1])])

# Se utiliza la función: mean square error para medir el error
# SGD define la forma de entrenar otorgando nuevos ajustes a los weights(parámetros internos) del modelo
red_modelo.compile(optimizer='sgd', loss='mean_squared_error')
```

Cada vez que un conjunto de datos pasa por un algoritmo, se dice que ha completado una **Epoch**.

Por lo tanto, **Epoch**, en el aprendizaje automático, se refiere al paso completo de datos de entrenamiento a través del algoritmo.

Es un hiperparámetro que determina el proceso de entrenamiento del modelo de aprendizaje automático

```
# Se procede a entrenar el modelo con un ciclo de 100 veces.
red_modelo.fit(datos_df['x'], datos_df['y'], epochs=100)
```

### **Predecir con nuevos datos**

Suponga que se tienen los siguientes puntos

```
nuevos_datos_df = pd.DataFrame([
    [6],
    [7],
    [8],
],
    columns=['x'])
nuevos_datos_df.head()
```

```
# La predicción puede consultarse de la siguiente manera
nuevos_datos_df['y'] = red_modelo.predict(nuevos_datos_df['x'])
```

## Presentación de resultados

```
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=datos_df['x'],
        y=datos_df['y'],
        mode='markers',
        name="Datos historicos",
        #line=go.scatter.Line(color="gray"),
        )
)

fig.add_trace(
    go.Scatter(
        x=nuevos_datos_df['x'],
        y=nuevos_datos_df['y'],
        mode='lines',
        name="Datos predichos",
        line=go.scatter.Line(color="gray"),
        )
)

fig.update_traces(marker_size=10)

fig.update_layout(
    title="Ejemplos de regresión lineal simple",
    xaxis_title="datos de variable independiente X",
    yaxis_title="datos de variable independiente y",
    legend_title="Legenda"
)

fig.show()
```
