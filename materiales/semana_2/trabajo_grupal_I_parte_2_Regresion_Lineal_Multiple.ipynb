## Resumen teórico
La **regresión lineal múltiple** modela la relación entre una variable respuesta continua $ y $ y varias variables explicativas $ X_1, X_2, ..., X_p $:

$ y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p + \varepsilon $

Donde:
- $ \beta_0 $: Intercepto (ventas esperadas sin inversión).
- $ \beta_i $: Efecto promedio de cada variable sobre las ventas.
- $ \varepsilon $: Error o ruido no explicado por el modelo.

Métricas comunes de evaluación:
- **MAE (Error Absoluto Medio)**
- **RMSE (Raíz del Error Cuadrático Medio)**
- **R² (Coeficiente de determinación)**

## 1. Carga de bibliotecas y datos
Ejecuta la siguiente celda para cargar las librerías y el dataset *Advertising*.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Dataset: Advertising.csv, incluido en esta carpeta.
df = pd.read_csv("Advertising.csv")
df.head()
```

## 2. Entrenamiento del modelo de regresión múltiple

```python
X = df[['TV', 'radio', 'newspaper']]
y = df['sales']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

modelo = LinearRegression()
modelo.fit(X_train, y_train)

y_pred = modelo.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print('Intercepto:', modelo.intercept_)
print('Coeficientes:', dict(zip(X.columns, modelo.coef_)))
print(f'R²={r2:.3f} | RMSE={rmse:.3f} | MAE={mae:.3f}')
```
