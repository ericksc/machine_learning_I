# Análisis del dataset `diabetes` y Modelado Lineal

Este notebook explora el dataset `diabetes` de scikit-learn y aplica un modelo de regresión lineal para predecir una variable objetivo basándose en las características del dataset.

## 3. Comparando características del dataset `diabetes`

Vamos a explorar otra característica del dataset `diabetes` y compararla con la que usamos en un análisis anterior (asumiendo que se usó una característica previamente, por ejemplo 'bmi') para ver cuál tiene un mejor coeficiente de determinación (R²).

Primero, cargamos el dataset.

**Uso de la comparación:** si elige variables observando estos resultados, el conjunto denominado `test` en este ejemplo actúa como validación. Para informar una evaluación final independiente, reserve un tercer conjunto antes de comparar.


```python
from sklearn.datasets import load_diabetes
import pandas as pd
```

```python
# Cargar el dataset
diabetes = load_diabetes()
X = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
y = diabetes.target
```

```python
# Mostrar las características disponibles
print("Características disponibles en el dataset diabetes:")
print(X.columns)
```

Ahora, seleccionaremos una característica diferente para compararla con 'bmi' (o la característica que hayas usado anteriormente). Por ejemplo, usemos 'bp' (presión sanguínea).

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np
```

```python
# Seleccionar las características
X_bmi = X[['bmi']]
X_bp = X[['bp']]
```

```python
# Dividir los datos en conjuntos de entrenamiento y prueba
X_train_bmi, X_test_bmi, y_train_bmi, y_test_bmi = train_test_split(X_bmi, y, test_size=0.2, random_state=42)
X_train_bp, X_test_bp, y_train_bp, y_test_bp = train_test_split(X_bp, y, test_size=0.2, random_state=42)
```

```python
# Crear y entrenar modelos de regresión lineal
model_bmi = LinearRegression()
model_bmi.fit(X_train_bmi, y_train_bmi)
```

```python
model_bp = LinearRegression()
model_bp.fit(X_train_bp, y_train_bp)
```

```python
# Realizar predicciones
y_pred_bmi = model_bmi.predict(X_test_bmi)
y_pred_bp = model_bp.predict(X_test_bp)

# Calcular el R² para ambos modelos
r2_bmi = r2_score(y_test_bmi, y_pred_bmi)
r2_bp = r2_score(y_test_bp, y_pred_bp)

print(f"R² para el modelo usando 'bmi': {r2_bmi:.4f}")
print(f"R² para el modelo usando 'bp': {r2_bp:.4f}")

# Determinar qué característica tiene mejor R²
if r2_bmi > r2_bp:
    print("'bmi' tiene un mejor R² que 'bp'.")
elif r2_bp > r2_bmi:
    print("'bp' tiene un mejor R² que 'bmi'.")
else:
    print("Ambas características tienen el mismo R².")
```

## 4. Gráfica de residuos y supuestos del modelo

Ahora vamos a graficar los residuos (la diferencia entre los valores reales y los predichos) para uno de los modelos (usaremos el de 'bmi' como ejemplo) y analizaremos si cumplen con los supuestos de linealidad y homocedasticidad.

Los **residuos** son la diferencia entre los valores observados de la variable dependiente y los valores predichos por el modelo de regresión. Un buen modelo lineal debe tener residuos que se distribuyan aleatoriamente alrededor de cero (linealidad) y que tengan una varianza constante a lo largo de los valores predichos (homocedasticidad).

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Calcular los residuos para el modelo 'bmi'
residuos_bmi = y_test_bmi - y_pred_bmi
```

```python
# Graficar los residuos frente a los valores predichos
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_pred_bmi, y=residuos_bmi)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel("Valores Predichos (bmi)")
plt.ylabel("Residuos (bmi)")
plt.title("Gráfico de Residuos vs. Valores Predichos (bmi)")
plt.grid(True)
plt.show()
```

Análisis de los supuestos:

- **Linealidad:** Si los residuos se distribuyen aleatoriamente alrededor de la línea horizontal en cero, sugiere que la relación entre la característica y la variable objetivo es lineal. Si hay un patrón curvilíneo, puede indicar una falta de linealidad.

- **Homocedasticidad:** Si la dispersión de los residuos es aproximadamente constante a lo largo de los valores predichos (la nube de puntos no se abre ni se cierra), sugiere homocedasticidad (varianza constante de los errores). Si la dispersión cambia, sugiere heterocedasticidad.

```python
# Gráfico de densidad de los residuos para verificar la normalidad.
plt.figure(figsize=(8, 5))
sns.histplot(residuos_bmi, kde=True)
plt.xlabel("Residuos (bmi)")
plt.ylabel("Frecuencia")
plt.title("Distribución de Residuos (bmi)")
plt.show()
```

**Conclusiones sobre los residuos:**

Observando el gráfico de residuos y la distribución de los residuos, podemos evaluar si se cumplen los supuestos de linealidad y homocedasticidad para el modelo basado en 'bmi'.

*   **Linealidad:** Verificar si los residuos parecen distribuirse aleatoriamente alrededor de cero o si hay un patrón.
*   **Homocedasticidad:** Verificar describe aquí si la dispersión de los residuos parece constante o si varía con los valores predichos.
