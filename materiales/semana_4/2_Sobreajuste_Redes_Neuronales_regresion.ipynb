# Sobreajuste en Redes Neuronales (Overfitting)

## Objetivo
Visualizar cómo se produce el **sobreajuste** en redes neuronales, comparando tres modelos con diferente complejidad sobre un dataset sintético.

---
## Teoría
El **sobreajuste (overfitting)** ocurre cuando un modelo aprende el ruido del conjunto de entrenamiento y pierde capacidad de generalizar.

**Síntomas:**
- Bajo error en entrenamiento, alto en validación.
- Divergencia de las curvas de pérdida.
- Modelo con demasiados parámetros para pocos datos.


```python
import tensorflow as tf
tf.keras.utils.set_random_seed(42)
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split

# Datos sintéticos: función cúbica con ruido
np.random.seed(42)
X = np.linspace(-1, 1, 100)
y = X**3 + np.random.randn(100)*0.1
X = X.reshape(-1, 1)

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.3, random_state=42)

plt.figure(figsize=(6,4))
plt.scatter(X_train, y_train, label='Entrenamiento', color='blue')
plt.scatter(X_val, y_val, label='Validación', color='red')
plt.title('Datos sintéticos (función cúbica con ruido)')
plt.legend()
plt.show()
```

```python
def train_and_plot(model, title, epochs=500):
    model.compile(optimizer=Adam(learning_rate=0.01), loss='mse')
    history = model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=epochs, verbose=0)

    # Curvas de pérdida
    plt.figure(figsize=(6,3))
    plt.plot(history.history['loss'], label='Entrenamiento')
    plt.plot(history.history['val_loss'], label='Validación')
    plt.title(f'{title} - Curvas de pérdida')
    plt.xlabel('Épocas')
    plt.ylabel('MSE')
    plt.legend()
    plt.show()

    # Predicciones
    X_plot = np.linspace(-1, 1, 100).reshape(-1, 1)
    y_pred = model.predict(X_plot)
    plt.figure(figsize=(6,4))
    plt.scatter(X_train, y_train, color='blue', label='Entrenamiento')
    plt.scatter(X_val, y_val, color='red', label='Validación')
    plt.plot(X_plot, y_pred, color='green', label='Predicción')
    plt.title(title)
    plt.legend()
    plt.show()
```

```python
# Modelo simple
model_simple = Sequential([
    Dense(4, activation='relu', input_shape=(1,)),
    Dense(1)
])
train_and_plot(model_simple, 'Modelo simple', epochs=300)
```

```python
# Modelo intermedio
model_medium = Sequential([
    Dense(32, activation='relu', input_shape=(1,)),
    Dense(16, activation='relu'),
    Dense(1)
])
train_and_plot(model_medium, 'Modelo intermedio', epochs=500)
```

```python
# Modelo complejo
model_complex = Sequential([
    Dense(128, activation='relu', input_shape=(1,)),
    Dense(128, activation='relu'),
    Dense(128, activation='relu'),
    Dense(1)
])
train_and_plot(model_complex, 'Modelo complejo', epochs=1000)
```
