# Técnicas de Clustering con K-Means y DBSCAN

## Objetivos de Aprendizaje
- Entender qué es el clustering y su propósito.

- Conocer las diferencias entre K-Means y DBSCAN.

- Aplicar ambas técnicas en Python usando scikit-learn.

- Visualizar y analizar los resultados.

## 1. ¿Qué es Clustering?
   
El clustering es una técnica de aprendizaje no supervisado que agrupa datos similares entre sí sin usar etiquetas previas.

Ejemplos de uso:

- Segmentación de clientes

- Agrupación de documentos o textos

- Detección de anomalías

- Agrupación de patrones en imágenes

## 2. Teoría de K-Means

### Definición
K-Means es un algoritmo que divide los datos en K grupos minimizando la distancia intra-cluster.

### Funcionamiento
1. Se eligen K centroides aleatorios.

2. Se asigna cada punto al centroide más cercano.

3. Se recalculan los centroides.

4. Repetir hasta convergencia.

## Limitaciones de K-Means
- Necesita que especifiques K.

- No funciona bien con formas no convexas o ruido.

- Sensible a valores atípicos.

# 3. Teoría de DBSCAN
### Definición
DBSCAN (Density-Based Spatial Clustering of Applications with Noise) agrupa puntos basándose en la densidad local.

- Conceptos clave
    - eps: radio para considerar vecinos.

    - min_samples: mínimo de puntos para formar una región densa.

- Clasifica puntos como:

    - Core: suficientes vecinos.

    - Border: pocos vecinos pero cerca de un core.

    - Noise: punto aislado.

## Ventajas
- No requiere número de clusters.

- Detección automática de outliers.

- Funciona bien con formas arbitrarias.

| Característica     | K-Means  | DBSCAN                       |
| ------------------ | -------- | ---------------------------- |
| Requiere K         | Sí     | No                         |
| Manejo de ruido    | Mal    | Bueno                      |
| Formas irregulares | Mal    | Bueno                      |
| Sensible a escala  | Sí     | Sí                         |
| Computacionalmente | Rápido | Más lento (pero flexible) |


## Buenas prácticas y consejos
- Estandariza tus datos antes de aplicar K-Means o DBSCAN.

- Para K-Means, usa el método del codo para elegir K.

- Para DBSCAN, prueba con diferentes valores de eps y min_samples.
