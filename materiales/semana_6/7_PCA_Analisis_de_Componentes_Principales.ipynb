# Análisis de Componentes Principales (PCA)


El **PCA (Principal Component Analysis)** es una técnica de **reducción de dimensionalidad** usada en análisis de datos y aprendizaje automático.
Permite reducir el número de variables en un conjunto de datos manteniendo la mayor cantidad posible de información.


Transforma las variables originales en **nuevas variables no correlacionadas** llamadas *componentes principales*.

## Cómo funciona el PCA
PCA usa álgebra lineal para transformar los datos en nuevas variables llamadas *componentes principales*.
Estas se obtienen calculando **vectores propios** (direcciones) y **valores propios** (importancia) de la matriz de covarianza.
Luego selecciona los componentes con los valores propios más grandes para proyectar los datos y conservar la mayor varianza posible.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

### Paso 1: Crear un conjunto de datos de ejemplo

```python
data = {
    'Altura': [170, 165, 180, 175, 160, 172, 168, 177, 162, 158],
    'Peso': [65, 59, 75, 68, 55, 70, 62, 74, 58, 54],
    'Edad': [30, 25, 35, 28, 22, 32, 27, 33, 24, 21],
    'Género': [1, 0, 1, 1, 0, 1, 0, 1, 0, 0]
}
df = pd.DataFrame(data)
df
```

Este escalador se utiliza para estandarizar las características, lo que significa que transformará los datos para que tengan una media de 0 y una desviación estándar de 1. Esto es importante para PCA, ya que es sensible a la escala de las variables.

```python
X = df.drop('Género', axis=1)
y = df['Género']
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0, stratify=y)
escalador = StandardScaler()
X_train_scaled = escalador.fit_transform(X_train_raw)
X_test_scaled = escalador.transform(X_test_raw)
# Proyección de todos los puntos exclusivamente para ilustrar el gráfico.
X_escalado = escalador.transform(X)
```

```python
pca = PCA(n_components=2)
X_train = pca.fit_transform(X_train_scaled)
X_test = pca.transform(X_test_scaled)
X_pca = pca.transform(X_escalado)
print("Varianza explicada:", pca.explained_variance_ratio_)
print("Ejemplo sintético de diez observaciones: no permite conclusiones sobre personas reales.")
```

```python
modelo = LogisticRegression()
modelo.fit(X_train, y_train)
y_pred = modelo.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Mujer', 'Hombre'], yticklabels=['Mujer', 'Hombre'])
plt.xlabel('Etiqueta predicha')
plt.ylabel('Etiqueta verdadera')
plt.title('Matriz de confusión')
plt.show()
```

```python
y_num = pd.factorize(y)[0]
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(X_escalado[:, 0], X_escalado[:, 1], c=y_num, cmap='coolwarm', edgecolor='k', s=80)
plt.xlabel('Característica original 1')
plt.ylabel('Característica original 2')
plt.title('Antes del PCA: primeras dos variables estandarizadas')
plt.colorbar(label='Clases objetivo')

plt.subplot(1, 2, 2)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_num, cmap='coolwarm', edgecolor='k', s=80)
plt.xlabel('Componente principal 1')
plt.ylabel('Componente principal 2')
plt.title('Después del PCA: proyección en 2 componentes principales')
plt.colorbar(label='Clases objetivo')
plt.tight_layout()
plt.show()
```

### Explicación de las gráficas

**Gráfico izquierdo (Antes del PCA):**  
Muestra los datos originales estandarizados utilizando las dos primeras características.  
No hay garantía de una separación clara entre clases, ya que se trata de dimensiones originales sin transformación.

**Gráfico derecho (Después del PCA):**  
Muestra los datos transformados usando los **dos componentes principales más importantes**.  
Estos nuevos componentes capturan la **máxima varianza**, lo que a menudo revela una mejor **separación entre clases** y una estructura más definida, facilitando el análisis o el modelado.

---

### Ventajas del Análisis de Componentes Principales (PCA)

- **Manejo de la multicolinealidad:**  
  Crea nuevas variables no correlacionadas para resolver problemas cuando las variables originales están fuertemente correlacionadas.

- **Reducción de ruido:**  
  Elimina los componentes con baja varianza, mejorando la claridad y calidad de los datos.

- **Compresión de datos:**  
  Representa la información con menos componentes, reduciendo el almacenamiento necesario y acelerando el procesamiento.

- **Detección de valores atípicos:**  
  Permite identificar puntos de datos inusuales al mostrar cuáles se desvían significativamente en el espacio reducido.

---

### Desventajas del PCA

- **Dificultad de interpretación:**  
  Los nuevos componentes son combinaciones lineales de las variables originales, lo que puede dificultar su explicación directa.

- **Sensibilidad al escalado de datos:**  
  Requiere una correcta estandarización antes de aplicarse; de lo contrario, los resultados pueden ser engañosos.

- **Pérdida de información:**  
  Al reducir dimensiones, puede perderse información importante si se conservan muy pocos componentes.

- **Suposición de linealidad:**  
  Funciona mejor cuando las relaciones entre variables son lineales; puede no capturar relaciones no lineales.

- **Complejidad computacional:**  
  En conjuntos de datos muy grandes, puede ser lento y consumir muchos recursos.

- **Riesgo de sobreajuste:**  
  Usar demasiados componentes o trabajar con conjuntos de datos pequeños puede generar modelos que no generalicen bien.
