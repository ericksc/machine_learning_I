# Trabajo Grupal II

Técnico de Análisis de datos.

Curso: Aprendizaje Automático I

Tema: Clusterización

Usando el archivo Iris.csv crear una solución para aplicar clusterización por medio de KMeans para descubir los grupos "Clusters" que se puede aplicar cada muestra "fila" del dataset.

Recordar aplicar el "elbow method" para encontrar el valor de k que más se ajusta.

Crear la lógica necesaria para:



*   Importar bibliotecas
*   Cargar el dataset
*   Hacer exploración de datos
*   Definir las caracteristicas
*   Aplicar escalamiento de las características
*   Aplicar el método de elbow
*   Aplicar clusterización con el valor de k selecionado.
*   Graficar para 2 características con leyenda para el valor del cluster.



Use las cuatro mediciones de sépalos y pétalos como características. Excluya identificadores y la columna de especie al ajustar K-Means. Puede usar la especie después, solo para contrastar los grupos; el número de clúster no es una etiqueta botánica.
